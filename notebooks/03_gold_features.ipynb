# Gold: características biomecánicas y temporales

Lee los Parquet de landmarks Silver, calcula características por frame con QC independiente por característica y las agrupa en ventanas no solapadas de 2 segundos. Se seleccionan frames cercanos a una rejilla de 5 FPS (sin interpolar landmarks); una característica solo se resume si tiene al menos 5 frames válidos en la ventana.

Genera `data/gold/plank_features.parquet`, `data/gold/hollow_body_features.parquet`, un diccionario JSON por ejercicio y observaciones de QC de características. Si no existen landmarks de hollow body, el dataset se crea vacío con el esquema previsto para facilitar la incorporación futura de datos.

```python
import logging
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'src').is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s %(levelname)s %(name)s: %(message)s',
)
WINDOW_SECONDS = 2.0
TARGET_FPS = 5.0
MIN_VALID_FRAMES = 5
```

```python
from src.pipeline import run_feature_engineering

gold_results = []
for exercise in ('plank', 'hollow_body'):
    result = run_feature_engineering(
        PROJECT_ROOT,
        exercise,
        window_seconds=WINDOW_SECONDS,
        target_fps=TARGET_FPS,
        min_valid_frames=MIN_VALID_FRAMES,
        overwrite=False,
    )
    gold_results.append(result)
gold_results
```
